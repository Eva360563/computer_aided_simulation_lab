```python
import math
import random
import itertools
from queue import PriorityQueue
from statistics import mean, stdev

# ==========================================
# 1. PARAMETRI
# ==========================================
SIM_TIME = 1440            # 24 ore in minuti
SOGLIA_ALLARME = 3         # sotto questa soglia di taxi per stand scatta il rebalancing (0 = spento)
MAX_ATTESA = 10.0          # minuti massimi di attesa del taxi prima della rinuncia
N_TAXI = 50
CAPIENZA_STAND = 10        # posti per stand
RAGGIO_STAND = 6.0         # distanza max (unità) per andare a piedi allo stand (600 m)
VEL_PEDONALE = 0.8         # unità/min (80 m/min = 4.8 km/h)
FATTORE_DOMANDA = 1.0      # moltiplicatore degli arrivi (per l'analisi di carico)
ATTESA_RIASSEGNAZIONE = None  # minuti allo stand dopo cui l'utente passa alla coda "app" (None = disattivata)

# Profilo orario: (inizio, fine, fascia) in minuti dalla mezzanotte
PROFILO = [
    (0, 420, "notte"),        # 00:00-07:00
    (420, 570, "punta"),      # 07:00-09:30
    (570, 1020, "giorno"),    # 09:30-17:00
    (1020, 1170, "punta"),    # 17:00-19:30
    (1170, 1260, "sera"),     # 19:30-21:00
    (1260, 1440, "notte"),    # 21:00-24:00
]
FASCE = ("notte", "punta", "giorno", "sera")
LEGENDA_FASCE = ("notte 00-07 e 21-24 | punta 07:00-09:30 e 17:00-19:30 | "
                 "giorno 09:30-17:00 | sera 19:30-21:00")
CANALI = ("app", "stand", "aeroporto")

# Per fascia: tempo medio tra arrivi (min) e velocità dei veicoli (1 unità = 100 m)
PARAMETRI_FASCIA = {
    "notte":  {"interarrivo": 5.0, "velocita": 6.5},   # 650 m/min = 39 km/h
    "punta":  {"interarrivo": 0.6, "velocita": 3.0},   # 300 m/min = 18 km/h
    "giorno": {"interarrivo": 1.0, "velocita": 4.0},   # 400 m/min = 24 km/h
    "sera":   {"interarrivo": 1.5, "velocita": 5.0},   # 500 m/min = 30 km/h
}
DURATA = {f: sum(fine - inizio for inizio, fine, n in PROFILO if n == f) for f in FASCE}

# Mobilità: origini = miscela di gaussiane (cx, cy, sigma, peso), pesi che sommano a 1
HOTSPOT = [
    (50, 50, 12, 0.40),   # centro
    (35, 35, 8, 0.20),    # zona università
    (55, 65, 8, 0.15),    # Porta Palazzo / mercato
    (55, 20, 10, 0.10),   # sud (Lingotto)
    (50, 50, 30, 0.15),   # periferia diffusa
]
LUNGHEZZA_MEDIA = 35.0            # lunghezza media corsa (Manhattan, unità = 3.5 km)
MIN_CORSA = 3.0                   # corsa minima (300 m)
FRAZIONE_VERSO_AEROPORTO = 0.03   # quota di utenti normali diretti a Caselle

# Domanda aeroportuale (arrivi a ondate)
VOLI_INIZIO = 360         # 06:00
VOLI_FINE = 1410          # 23:30
INTERVALLO_VOLI = 25.0    # minuti medi tra due voli
PAX_TAXI_PER_VOLO = 4.0   # passeggeri medi per volo che cercano un taxi

STATI = ("idle", "assigned", "busy", "returning")

stand_coords = {
    "Caselle": (10, 90),        # aeroporto
    "Susa": (40, 50),
    "Nuova": (50, 40),
    "Castello": (50, 50),
    "Vittorio": (60, 50),
    "Lingotto": (55, 15),
    "Politecnico": (35, 35),
    "Porta_Palazzo": (55, 65),
}
STAND_AEROPORTO = "Caselle"

# ==========================================
# 2. FUNZIONI DI SUPPORTO
# ==========================================
def manhattan(x1, y1, x2, y2):
    return abs(x1 - x2) + abs(y1 - y2)

def fascia(t):
    for inizio, fine, nome in PROFILO:
        if inizio <= t < fine:
            return nome
    return PROFILO[-1][2]

def velocita(t):
    # velocità in vigore all'istante di partenza del viaggio
    return PARAMETRI_FASCIA[fascia(t)]["velocita"]

def ripartisci(t0, t1):
    """Divide l'intervallo [t0, t1] tra i minuti di ciascuna fascia."""
    out = dict.fromkeys(FASCE, 0.0)
    for inizio, fine, nome in PROFILO:
        out[nome] += max(0.0, min(t1, fine) - max(t0, inizio))
    return out

def poisson(rng, lam):
    if lam <= 0:
        return 0
    soglia = math.exp(-lam)
    k, p = 0, 1.0
    while True:
        p *= rng.random()
        if p <= soglia:
            return k
        k += 1

def campiona_origine(rng):
    """Punto della mappa dalla miscela di gaussiane (campionamento per rifiuto)."""
    r, acc = rng.random(), 0.0
    for cx, cy, sigma, peso in HOTSPOT:
        acc += peso
        if r <= acc:
            break
    while True:
        x, y = rng.gauss(cx, sigma), rng.gauss(cy, sigma)
        if 0 <= x <= 100 and 0 <= y <= 100:
            return x, y

def campiona_destinazione(rng, x0, y0):
    """Lunghezza corsa esponenziale (decadimento della distanza), direzione casuale."""
    if rng.random() < FRAZIONE_VERSO_AEROPORTO:
        return stand_coords[STAND_AEROPORTO]
    for _ in range(50):
        L = max(MIN_CORSA, rng.expovariate(1.0 / LUNGHEZZA_MEDIA))
        f = rng.random()
        x = x0 + f * L * rng.choice((-1, 1))
        y = y0 + (1 - f) * L * rng.choice((-1, 1))
        if 0 <= x <= 100 and 0 <= y <= 100:
            return x, y
    return min(100, max(0, x)), min(100, max(0, y))

# ==========================================
# 3. ENTITÀ E MISURE
# ==========================================
class Utente:
    def __init__(self, id_utente, tempo_chiamata, origine, destinazione, canale="app"):
        self.id = id_utente
        self.tempo_chiamata = tempo_chiamata
        self.fascia = fascia(tempo_chiamata)   # fascia in cui l'utente è comparso
        self.canale = canale                   # "app", "stand", "aeroporto"
        self.stato = "in attesa"  # "camminando", "in attesa", "assegnato", "servito", "rinunciato"
        self.x_orig, self.y_orig = origine
        self.x_dest, self.y_dest = destinazione
        self.x_pick, self.y_pick = origine     # punto in cui sale sul taxi
        self.stand = None          # stand raggiunto a piedi (None = usa l'app)
        self.camminata = 0.0       # minuti di camminata
        self.tempo_pronto = tempo_chiamata  # da qui parte l'attesa del taxi

class Taxi:
    def __init__(self, id_taxi, x, y, stand=None):
        self.id = id_taxi
        self.x = x
        self.y = y
        self.stato = "idle"   # "idle", "assigned", "busy", "returning"
        self.stand = stand    # stand in cui è parcheggiato (None = fermo fuori stand)
        self.target = None    # stand di destinazione durante il rientro/rebalancing
        self.t_stato = 0.0    # istante dell'ultimo cambio di stato

class Measure:
    def __init__(self, n_taxi):
        self.n_taxi = n_taxi
        self.generati = dict.fromkeys(FASCE, 0)
        self.aeroporto = dict.fromkeys(FASCE, 0)           # utenti da voli
        self.serviti = dict.fromkeys(FASCE, 0)
        self.persi = dict.fromkeys(FASCE, 0)
        self.attesa_totale = dict.fromkeys(FASCE, 0.0)
        self.da_stand = dict.fromkeys(FASCE, 0)            # utenti che camminano allo stand
        self.camminata_totale = dict.fromkeys(FASCE, 0.0)
        self.viaggi_a_vuoto = dict.fromkeys(FASCE, 0)
        self.tempo = {s: dict.fromkeys(FASCE, 0.0) for s in STATI}  # minuti-taxi per stato
        # misure per canale
        self.gen_c = dict.fromkeys(CANALI, 0)
        self.serv_c = dict.fromkeys(CANALI, 0)
        self.persi_c = dict.fromkeys(CANALI, 0)
        self.attesa_c = dict.fromkeys(CANALI, 0.0)

# ==========================================
# 4. SIMULATORE
# ==========================================
class Simulazione:
    def __init__(self, seed=None, n_taxi=N_TAXI, fattore_domanda=FATTORE_DOMANDA,
                 soglia=SOGLIA_ALLARME, attesa_riassegnazione=ATTESA_RIASSEGNAZIONE):
        self.rng = random.Random(seed)
        self.FES = PriorityQueue()
        self.seq = itertools.count()   # tie-breaker per eventi simultanei
        self.m = Measure(n_taxi)
        self.fattore = fattore_domanda
        self.soglia = soglia
        self.attesa_riass = attesa_riassegnazione
        self.lambda_max = fattore_domanda * max(1.0 / p["interarrivo"] for p in PARAMETRI_FASCIA.values())
        self.id_counter = 0
        self.code_stand = {nome: [] for nome in stand_coords}
        self.coda_app = []
        self.flotta = []

        # posizionamento iniziale rispettando la capienza degli stand
        occupati = dict.fromkeys(stand_coords, 0)
        for i in range(n_taxi):
            liberi = [n for n in stand_coords if occupati[n] < CAPIENZA_STAND]
            if liberi:
                n = self.rng.choice(liberi)
                occupati[n] += 1
                self.flotta.append(Taxi(i, *stand_coords[n], stand=n))
            else:
                # flotta più grande dei posti: taxi fermo fuori stand
                n = self.rng.choice(list(stand_coords))
                self.flotta.append(Taxi(i, *stand_coords[n], stand=None))

    def schedula(self, t, tipo, taxi=None, dato=None):
        self.FES.put((t, next(self.seq), tipo, taxi, dato))

    def tasso_arrivo(self, t):
        return self.fattore / PARAMETRI_FASCIA[fascia(t)]["interarrivo"]

    def prossimo_arrivo(self, t):
        """Thinning: processo di Poisson non omogeneo con tasso costante a tratti."""
        while True:
            t += self.rng.expovariate(self.lambda_max)
            if t > SIM_TIME:
                return None
            if self.rng.random() < self.tasso_arrivo(t) / self.lambda_max:
                return t

    def prossimo_volo(self, t):
        t += self.rng.expovariate(1.0 / INTERVALLO_VOLI)
        return t if t <= VOLI_FINE else None

    # ---------- utilità ----------
    def set_stato(self, t, taxi, nuovo):
        """Cambia stato al taxi accumulando il tempo trascorso nello stato precedente."""
        for f, minuti in ripartisci(taxi.t_stato, t).items():
            self.m.tempo[taxi.stato][f] += minuti
        taxi.stato = nuovo
        taxi.t_stato = t

    def posti_occupati(self, nome):
        # taxi parcheggiati + taxi in arrivo (il posto è già prenotato)
        return sum(
            1 for x in self.flotta
            if (x.stato == "idle" and x.stand == nome)
            or (x.stato == "returning" and x.target == nome)
        )

    def taxi_allo_stand(self, nome):
        for x in self.flotta:
            if x.stato == "idle" and x.stand == nome:
                return x
        return None

    def assegna(self, t, taxi, utente, tempo_pickup):
        self.set_stato(t, taxi, "assigned")
        taxi.stand = None          # il taxi lascia lo stand
        utente.stato = "assegnato"
        self.schedula(t + tempo_pickup, "inizio_servizio", taxi, utente)

    def servi_da_stand(self, t, taxi, nome):
        utente = self.code_stand[nome].pop(0)
        self.assegna(t, taxi, utente, 0.0)

    def smista_coda_app(self, t):
        """Assegna i taxi idle agli utenti in coda 'app' (FIFO), dal punto di raccolta."""
        for u in list(self.coda_app):
            idle = [x for x in self.flotta if x.stato == "idle"]
            if not idle:
                return
            taxi = min(idle, key=lambda x: manhattan(u.x_pick, u.y_pick, x.x, x.y))
            pickup = manhattan(u.x_pick, u.y_pick, taxi.x, taxi.y) / velocita(t)
            if t + pickup > u.tempo_pronto + MAX_ATTESA:
                continue  # troppo lontano: l'utente rinuncerà o aspetterà un altro taxi
            self.coda_app.remove(u)
            self.assegna(t, taxi, u, pickup)

    def ritorna_a_stand(self, t, taxi):
        """Fleet return + rebalancing: sceglie lo stand con posti liberi."""
        occupati = {n: self.posti_occupati(n) for n in stand_coords}
        liberi = [n for n in stand_coords if occupati[n] < CAPIENZA_STAND]
        if not liberi:
            return  # stand tutti pieni: il taxi resta idle dov'è (assegnabile via app)

        def dist(n):
            return manhattan(taxi.x, taxi.y, *stand_coords[n])

        in_crisi = [n for n in liberi if occupati[n] < self.soglia]
        piu_vicino = min(liberi, key=dist)
        dest = min(in_crisi, key=dist) if in_crisi else piu_vicino
        if dest != piu_vicino:
            self.m.viaggi_a_vuoto[fascia(t)] += 1   # rebalancing vero, non semplice rientro

        self.set_stato(t, taxi, "returning")
        taxi.stand = None
        taxi.target = dest
        self.schedula(t + dist(dest) / velocita(t), "arrivo_stand", taxi, dest)

    # ---------- eventi ----------
    def arrivo_utente(self, t, _taxi, _dato):
        self.id_counter += 1
        prossimo = self.prossimo_arrivo(t)
        if prossimo is not None:
            self.schedula(prossimo, "arrivo_utente")

        origine = campiona_origine(self.rng)
        destinazione = campiona_destinazione(self.rng, *origine)
        u = Utente(self.id_counter, t, origine, destinazione)
        self.m.generati[u.fascia] += 1

        nome, dist = min(
            ((n, manhattan(u.x_orig, u.y_orig, sx, sy)) for n, (sx, sy) in stand_coords.items()),
            key=lambda c: c[1],
        )
        if dist <= RAGGIO_STAND:
            # va a piedi allo stand più vicino
            u.canale = "stand"
            self.m.gen_c["stand"] += 1
            u.stand = nome
            u.camminata = dist / VEL_PEDONALE
            u.stato = "camminando"
            self.m.da_stand[u.fascia] += 1
            self.m.camminata_totale[u.fascia] += u.camminata
            self.schedula(t + u.camminata, "utente_allo_stand", dato=u)
        else:
            # usa l'app
            u.canale = "app"
            self.m.gen_c["app"] += 1
            u.tempo_pronto = t
            self.schedula(t + MAX_ATTESA, "rinuncia_utente", dato=u)
            self.coda_app.append(u)
            self.smista_coda_app(t)

    def arrivo_volo(self, t, _taxi, _dato):
        prossimo = self.prossimo_volo(t)
        if prossimo is not None:
            self.schedula(prossimo, "arrivo_volo")

        for _ in range(poisson(self.rng, PAX_TAXI_PER_VOLO * self.fattore)):
            self.id_counter += 1
            origine = stand_coords[STAND_AEROPORTO]
            destinazione = campiona_origine(self.rng)   # destinazione in città
            u = Utente(self.id_counter, t, origine, destinazione, canale="aeroporto")
            u.stand = STAND_AEROPORTO                   # nessuna camminata
            self.m.generati[u.fascia] += 1
            self.m.aeroporto[u.fascia] += 1
            self.m.gen_c["aeroporto"] += 1
            self.schedula(t + MAX_ATTESA, "rinuncia_utente", dato=u)
            if self.attesa_riass is not None:
                self.schedula(t + self.attesa_riass, "ripiega_su_app", dato=u)
            self.code_stand[STAND_AEROPORTO].append(u)
            taxi = self.taxi_allo_stand(STAND_AEROPORTO)
            if taxi:
                self.servi_da_stand(t, taxi, STAND_AEROPORTO)

    def utente_allo_stand(self, t, _taxi, utente):
        utente.tempo_pronto = t
        utente.stato = "in attesa"
        utente.x_pick, utente.y_pick = stand_coords[utente.stand]   # sale allo stand
        self.schedula(t + MAX_ATTESA, "rinuncia_utente", dato=utente)
        if self.attesa_riass is not None:
            self.schedula(t + self.attesa_riass, "ripiega_su_app", dato=utente)
        self.code_stand[utente.stand].append(utente)
        taxi = self.taxi_allo_stand(utente.stand)
        if taxi:
            self.servi_da_stand(t, taxi, utente.stand)

    def ripiega_su_app(self, t, _taxi, utente):
        """Chi aspetta allo stand da troppo passa alla coda dei clienti serviti dal taxi libero più vicino."""
        if utente.stato == "in attesa" and utente in self.code_stand[utente.stand]:
            self.code_stand[utente.stand].remove(utente)
            self.coda_app.append(utente)
            self.smista_coda_app(t)

    def inizio_servizio(self, t, taxi, utente):
        utente.stato = "servito"
        self.set_stato(t, taxi, "busy")
        self.m.serviti[utente.fascia] += 1
        self.m.attesa_totale[utente.fascia] += t - utente.tempo_pronto
        self.m.serv_c[utente.canale] += 1
        self.m.attesa_c[utente.canale] += t - utente.tempo_pronto

        tempo_viaggio = manhattan(utente.x_pick, utente.y_pick, utente.x_dest, utente.y_dest) / velocita(t)
        taxi.x, taxi.y = utente.x_dest, utente.y_dest  # posizione a fine corsa (taxi occupato, non assegnabile)
        self.schedula(t + tempo_viaggio, "dropoff", taxi)

    def dropoff(self, t, taxi, _dato):
        self.set_stato(t, taxi, "idle")
        taxi.stand = None
        self.smista_coda_app(t)          # prima prova a servire chi aspetta in coda 'app'
        if taxi.stato == "idle":         # se nessuno lo ha preso, torna a uno stand
            self.ritorna_a_stand(t, taxi)

    def arrivo_stand(self, t, taxi, nome):
        taxi.x, taxi.y = stand_coords[nome]
        self.set_stato(t, taxi, "idle")
        taxi.stand = nome
        taxi.target = None
        if self.code_stand[nome]:
            self.servi_da_stand(t, taxi, nome)
        else:
            self.smista_coda_app(t)

    def rinuncia_utente(self, t, _taxi, utente):
        if utente.stato == "in attesa":
            utente.stato = "rinunciato"
            self.m.persi[utente.fascia] += 1
            self.m.persi_c[utente.canale] += 1
            if utente in self.coda_app:
                self.coda_app.remove(utente)
            for coda in self.code_stand.values():
                if utente in coda:
                    coda.remove(utente)

    # ---------- loop ----------
    def run(self):
        primo = self.prossimo_arrivo(0.0)
        if primo is not None:
            self.schedula(primo, "arrivo_utente")
        primo_volo = self.prossimo_volo(VOLI_INIZIO)
        if primo_volo is not None:
            self.schedula(primo_volo, "arrivo_volo")

        handlers = {
            "arrivo_utente": self.arrivo_utente,
            "arrivo_volo": self.arrivo_volo,
            "utente_allo_stand": self.utente_allo_stand,
            "ripiega_su_app": self.ripiega_su_app,
            "inizio_servizio": self.inizio_servizio,
            "dropoff": self.dropoff,
            "arrivo_stand": self.arrivo_stand,
            "rinuncia_utente": self.rinuncia_utente,
        }
        while not self.FES.empty():
            t, _, tipo, taxi, dato = self.FES.get()
            if t > SIM_TIME:
                break
            handlers[tipo](t, taxi, dato)

        # chiusura: accumula l'ultimo tratto di stato di ogni taxi
        for taxi in self.flotta:
            for f, minuti in ripartisci(taxi.t_stato, SIM_TIME).items():
                self.m.tempo[taxi.stato][f] += minuti
        return self.m

# ==========================================
# 5. KPI E OUTPUT
# ==========================================
def _calcola(m, fasce):
    gen = sum(m.generati[f] for f in fasce)
    aer = sum(m.aeroporto[f] for f in fasce)
    serv = sum(m.serviti[f] for f in fasce)
    persi = sum(m.persi[f] for f in fasce)
    attesa = sum(m.attesa_totale[f] for f in fasce)
    stand = sum(m.da_stand[f] for f in fasce)
    cammin = sum(m.camminata_totale[f] for f in fasce)
    vuoti = sum(m.viaggi_a_vuoto[f] for f in fasce)
    minuti_taxi = m.n_taxi * sum(DURATA[f] for f in fasce)

    def quota_tempo(stato):
        return 100 * sum(m.tempo[stato][f] for f in fasce) / minuti_taxi

    gestiti = serv + persi
    gen_citta = gen - aer
    return {
        "generati": gen,
        "aeroporto": aer,
        "serviti": serv,
        "persi": persi,
        "in_corso": gen - gestiti,
        "disponibilita": 100 * serv / gestiti if gestiti else 0.0,
        "attesa_media": attesa / serv if serv else 0.0,
        "quota_stand": 100 * stand / gen_citta if gen_citta else 0.0,
        "camminata_media": cammin / stand if stand else 0.0,
        "utilizzo": quota_tempo("busy"),
        "pickup": quota_tempo("assigned"),
        "a_vuoto": quota_tempo("returning"),
        "idle": quota_tempo("idle"),
        "viaggi_a_vuoto": vuoti,
    }

def _calcola_canale(m, c):
    serv, persi = m.serv_c[c], m.persi_c[c]
    gestiti = serv + persi
    return {
        "generati": m.gen_c[c],
        "serviti": serv,
        "persi": persi,
        "disponibilita": 100 * serv / gestiti if gestiti else 0.0,
        "attesa_media": m.attesa_c[c] / serv if serv else 0.0,
    }

def kpi(m):
    """KPI per ciascuna fascia, 'totale' e 'canali' (app/stand/aeroporto)."""
    out = {f: _calcola(m, (f,)) for f in FASCE}
    out["totale"] = _calcola(m, FASCE)
    out["canali"] = {c: _calcola_canale(m, c) for c in CANALI}
    return out

def esegui(n_rep, **kwargs):
    return [kpi(Simulazione(seed=s, **kwargs).run()) for s in range(n_rep)]

CAMPI = [
    ("generati", "Utenti generati"),
    ("aeroporto", "  di cui da voli (Caselle)"),
    ("disponibilita", "Disponibilità (%)"),
    ("attesa_media", "Attesa media taxi (min)"),
    ("camminata_media", "Camminata media (min)"),
    ("quota_stand", "Utenti allo stand (%, esclusi voli)"),
    ("utilizzo", "Utilizzo: taxi con passeggero (%)"),
    ("pickup", "Taxi in pick-up (%)"),
    ("a_vuoto", "Taxi in rientro/rebalancing (%)"),
    ("idle", "Taxi fermi (%)"),
    ("viaggi_a_vuoto", "Viaggi a vuoto per rebalancing"),
]
CAMPI_CANALE = [
    ("generati", "Utenti generati"),
    ("serviti", "Serviti"),
    ("persi", "Persi"),
    ("disponibilita", "Disponibilità (%)"),
    ("attesa_media", "Attesa media taxi (min)"),
]
COLONNE = ("totale",) + FASCE

# valori t di Student al 95% (bilaterale) per df = 1..20
T95 = {1: 12.706, 2: 4.303, 3: 3.182, 4: 2.776, 5: 2.571, 6: 2.447, 7: 2.365, 8: 2.306,
       9: 2.262, 10: 2.228, 11: 2.201, 12: 2.179, 13: 2.160, 14: 2.145, 15: 2.131,
       16: 2.120, 17: 2.110, 18: 2.101, 19: 2.093, 20: 2.086}

def media_ic(vals):
    """Media e semi-ampiezza dell'intervallo di confidenza al 95%."""
    n = len(vals)
    t = T95.get(n - 1, 2.0 if n < 60 else 1.96)
    return mean(vals), t * stdev(vals) / math.sqrt(n)

def fmt_ic(vals):
    m, h = media_ic(vals)
    return f"{m:.1f} ± {h:.1f}"

def riga(etichetta, celle, larghezza):
    return f"{etichetta:<38s}" + "".join(f"{c:>{larghezza}s}" for c in celle)

def stampa_run(k):
    print(riga("", COLONNE, 12))
    for campo, etichetta in CAMPI:
        print(riga(etichetta, [f"{k[c][campo]:.2f}" for c in COLONNE], 12))
    r = k["totale"]
    print(f"\nServiti: {r['serviti']} | Persi: {r['persi']} | Ancora in corso a fine sim: {r['in_corso']}")

def stampa_canali(k):
    print(riga("PER CANALE", CANALI, 12))
    for campo, etichetta in CAMPI_CANALE:
        print(riga(etichetta, [f"{k['canali'][c][campo]:.2f}" for c in CANALI], 12))

def stampa_repliche(risultati):
    print(riga("", COLONNE, 17))
    for campo, etichetta in CAMPI:
        celle = []
        for c in COLONNE:
            vals = [r[c][campo] for r in risultati]
            celle.append(f"{mean(vals):.2f} ± {stdev(vals):.2f}")
        print(riga(etichetta, celle, 17))

def stampa_canali_repliche(risultati):
    print(riga("PER CANALE", CANALI, 17))
    for campo, etichetta in CAMPI_CANALE:
        celle = []
        for c in CANALI:
            vals = [r["canali"][c][campo] for r in risultati]
            celle.append(f"{mean(vals):.2f} ± {stdev(vals):.2f}")
        print(riga(etichetta, celle, 17))

def sweep(flotte, fattori, n_rep):
    """Confronto flotta x carico di domanda, con IC al 95% sulle repliche."""
    print(f"{'Flotta':>7}{'Domanda':>9}{'Disp. tot (%)':>19}{'Disp. punta (%)':>19}"
          f"{'Attesa (min)':>17}{'Util. punta (%)':>17}{'Viaggi vuoto':>14}")
    for n in flotte:
        for f in fattori:
            ris = esegui(n_rep, n_taxi=n, fattore_domanda=f)
            d_tot = media_ic([r["totale"]["disponibilita"] for r in ris])
            d_pun = media_ic([r["punta"]["disponibilita"] for r in ris])
            att = media_ic([r["totale"]["attesa_media"] for r in ris])
            u_pun = mean(r["punta"]["utilizzo"] for r in ris)
            vuoti = mean(r["totale"]["viaggi_a_vuoto"] for r in ris)
            print(f"{n:>7}{'x' + str(f):>9}"
                  f"{f'{d_tot[0]:.1f} ± {d_tot[1]:.1f}':>19}"
                  f"{f'{d_pun[0]:.1f} ± {d_pun[1]:.1f}':>19}"
                  f"{f'{att[0]:.2f} ± {att[1]:.2f}':>17}"
                  f"{u_pun:>17.1f}{vuoti:>14.0f}")

def sweep_soglia(soglie, n_taxi, fattore, n_rep):
    """Effetto della soglia di rebalancing (0 = nessun rebalancing) a flotta e domanda fissate."""
    print(f"Flotta {n_taxi}, domanda x{fattore}, {n_rep} repliche, IC 95%")
    print(f"{'Soglia':>7}{'Disp. tot (%)':>19}{'Disp. punta (%)':>19}{'Attesa (min)':>17}"
          f"{'Rientro punta (%)':>19}{'Viaggi vuoto':>14}")
    for s in soglie:
        ris = esegui(n_rep, n_taxi=n_taxi, fattore_domanda=fattore, soglia=s)
        d_tot = media_ic([r["totale"]["disponibilita"] for r in ris])
        d_pun = media_ic([r["punta"]["disponibilita"] for r in ris])
        att = media_ic([r["totale"]["attesa_media"] for r in ris])
        rient = mean(r["punta"]["a_vuoto"] for r in ris)
        vuoti = mean(r["totale"]["viaggi_a_vuoto"] for r in ris)
        print(f"{s:>7}"
              f"{f'{d_tot[0]:.1f} ± {d_tot[1]:.1f}':>19}"
              f"{f'{d_pun[0]:.1f} ± {d_pun[1]:.1f}':>19}"
              f"{f'{att[0]:.2f} ± {att[1]:.2f}':>17}"
              f"{rient:>19.1f}{vuoti:>14.0f}")

def confronto_riassegnazione(opzioni, n_taxi, fattore, n_rep):
    """Confronta il caso base (None) con la riassegnazione degli utenti in attesa agli stand."""
    print(f"Flotta {n_taxi}, domanda x{fattore}, {n_rep} repliche, IC 95%")
    print(f"{'Riass. dopo':>12}{'Disp. app':>16}{'Disp. stand':>16}{'Disp. aeroporto':>18}"
          f"{'Disp. totale':>16}{'Attesa (min)':>16}")
    for x in opzioni:
        ris = esegui(n_rep, n_taxi=n_taxi, fattore_domanda=fattore, attesa_riassegnazione=x)
        etichetta = "nessuna" if x is None else f"{x} min"
        print(f"{etichetta:>12}"
              f"{fmt_ic([r['canali']['app']['disponibilita'] for r in ris]):>16}"
              f"{fmt_ic([r['canali']['stand']['disponibilita'] for r in ris]):>16}"
              f"{fmt_ic([r['canali']['aeroporto']['disponibilita'] for r in ris]):>18}"
              f"{fmt_ic([r['totale']['disponibilita'] for r in ris]):>16}"
              f"{media_ic([r['totale']['attesa_media'] for r in ris])[0]:>16.2f}")

if __name__ == "__main__":
    ESEGUI_SWEEP_FLOTTA = True
    ESEGUI_SWEEP_SOGLIA = True
    ESEGUI_CONFRONTO_RIASSEGNAZIONE = True
    N_REP = 20
    N_REP_SWEEP = 10

    print(f"Fasce: {LEGENDA_FASCE}\n")

    print("=== SINGOLA RUN (seed=42) ===")
    k = kpi(Simulazione(seed=42).run())
    stampa_run(k)
    print()
    stampa_canali(k)

    print(f"\n=== MEDIA SU {N_REP} REPLICHE (media ± dev. std.) ===")
    ris = esegui(N_REP)
    stampa_repliche(ris)
    print()
    stampa_canali_repliche(ris)

    if ESEGUI_SWEEP_FLOTTA:
        print(f"\n=== CONFRONTO FLOTTA x DOMANDA ({N_REP_SWEEP} repliche, IC 95%) ===")
        sweep((30, 50, 80), (1.0, 1.5, 2.0), N_REP_SWEEP)

    if ESEGUI_SWEEP_SOGLIA:
        print("\n=== SWEEP SOGLIA DI REBALANCING (0 = spento) ===")
        sweep_soglia((0, 1, 2, 3, 4, 5), n_taxi=50, fattore=1.5, n_rep=N_REP_SWEEP)

    if ESEGUI_CONFRONTO_RIASSEGNAZIONE:
        print("\n=== RIASSEGNAZIONE DEGLI UTENTI IN ATTESA AGLI STAND ===")
        confronto_riassegnazione((None, 2.0, 4.0), n_taxi=50, fattore=1.5, n_rep=N_REP_SWEEP)
```
